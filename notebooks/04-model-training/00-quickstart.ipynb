# model training

**propose:** fine-tune base language models using supervised fine-tuning loops. load model architecture, pack hyperparameters, and run training with checkpoint saves. import model building and training harness from `src.models/`.

**input:** tokenized datasets from `data/processed/` (train/val) and training config from `configs/default.yaml`.

**output:** trained model adapters/checkpoints saved to `artifacts/training/checkpoints/` and export to `artifacts/training/models/`.
